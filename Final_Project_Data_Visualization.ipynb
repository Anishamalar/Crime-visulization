***Analytical Question***




Question 1: Which police districts in Chicago have the highest and lowest arrest rates for different types of crimes?

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
district_arrest_rates = df.groupby(['District', 'Primary Type'])['Arrest'].mean().reset_index()
district_arrest_rates['Rank'] = district_arrest_rates.groupby('Primary Type')['Arrest'].rank(ascending=False)
highest_arrest_rates = district_arrest_rates[district_arrest_rates['Rank'] == 1]
district_arrest_rates['Rank'] == district_arrest_rates.groupby('Primary Type')['Arrest'].transform('max').rank(ascending=False)
plt.figure(figsize=(20, 20))
plt.subplot(2, 1, 1)
sns.barplot(x='District', y='Arrest', hue='Primary Type', data=highest_arrest_rates)
plt.title('Districts with Highest Arrest Rates for Different Crime Types')
plt.xlabel('Police District')
plt.ylabel('Arrest Rate')
plt.xticks(rotation=45)
plt.legend(title='Crime Type')
print("Districts with Highest Arrest Rates for Different Crime Types:")
print(highest_arrest_rates[['District', 'Primary Type', 'Arrest']])

plt.subplot(2, 1, 2)
sns.barplot(x='District', y='Arrest', hue='Primary Type', data=lowest_arrest_rates)
plt.title('Districts with Lowest Arrest Rates for Different Crime Types')
plt.xlabel('Police District')
plt.ylabel('Arrest Rate')
plt.xticks(rotation=45)
plt.legend(title='Crime Type')
print("\nDistricts with Lowest Arrest Rates for Different Crime Types:")
print(lowest_arrest_rates[['District', 'Primary Type', 'Arrest']])
plt.tight_layout()
plt.show()
```

We can observe from the graph and values that the police districts 15.0 and 21.0 have the highest number of crimes like homicide, public peace violation , stalking and weapons violation. Districts with lowest crime are 14.0 and 20.0.

 Question 2 : Is there a correlation between the day of the week and the number of reported crimes? How does this correlation vary by crime type?

```python
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y %H:%M', errors='coerce')
df.dropna(subset=['Date'], inplace=True)
df['DayOfWeek'] = df['Date'].dt.dayofweek
```

```python
crime_counts = df.groupby(['DayOfWeek', 'Primary Type']).size().reset_index(name='CrimeCount')
avg_crime_counts = crime_counts.groupby('DayOfWeek')['CrimeCount'].mean().reset_index()
plt.figure(figsize=(20, 10))
sns.barplot(x='DayOfWeek', y='CrimeCount', hue='Primary Type', data=crime_counts, ci=None)
plt.title('Number of Reported Crimes by Day of the Week and Crime Type')
plt.xlabel('Day of the Week')
plt.ylabel('Number of Reported Crimes')
plt.xticks(rotation=45)
plt.legend(title='Crime Type')
plt.tight_layout()
plt.show()
pd.set_option('display.max_rows', None)
print("Number of Reported Crimes by Day of the Week and Crime Type:")
print(crime_counts)
```

Yes, there is a correlation between reported crimes and day of the week . We can observe from the graph and printed values that crimes like battery and assault  happen more during weekends than weekdays.

Question 3:How does the likelihood of arrest vary across different types of crimes ?

```python
crime_counts = df.groupby(['Primary Type', 'Arrest']).size().reset_index(name='Incidents')
crime_counts_pivot = crime_counts.pivot_table(index='Primary Type', columns='Arrest', values='Incidents', aggfunc='sum', fill_value=0)
plt.figure(figsize=(20, 10))
crime_counts_pivot.plot(kind='bar', width=0.8, color=['skyblue', 'salmon'])
plt.title('Arrests vs. Non-Arrests by Crime Type')
plt.xlabel('Crime Type')
plt.ylabel('Number of Incidents')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Arrest', labels=['Not Arrested', 'Arrested'])
plt.tight_layout()
plt.show()

print("Arrests vs. Non-Arrests by Crime Type:")
print(crime_counts_pivot)
```

We can observe from the chart and printed values that many cases were filed for battery , criminal damage and deceptive practices. But in most of the cases they were not arrested. Battery , weapons violation and  Narcotics have the higher number of arrests cases.

Question:4 How do arrest/non-arrest and domestic/non-domestic incidents vary across different Location descriptions in Chicago?

```python
top_location_descriptions = df['Location Description'].value_counts().index[:5]
df = df[df['Location Description'].isin(top_location_descriptions)]
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(16, 12))
sns.countplot(x='Location Description', hue='Arrest', data=df, palette='Set2', hue_order=[True, False], ax=axes[0, 0])
axes[0, 0].set_title('Arrests vs. Non-Arrests by Location Description')
axes[0, 0].set_xlabel('Location Description')
axes[0, 0].set_ylabel('Number of Incidents')
axes[0, 0].legend(title='Arrest', labels=['Arrested', 'Not Arrested'])
axes[0, 0].tick_params(axis='x', rotation=45)

sns.countplot(x='Location Description', hue='Domestic', data=df, palette='Set2', hue_order=[True, False], ax=axes[0, 1])
axes[0, 1].set_title('Domestic vs. Non-Domestic Incidents by Location Description')
axes[0, 1].set_xlabel('Location Description')
axes[0, 1].set_ylabel('Number of Incidents')
axes[0, 1].legend(title='Domestic', labels=['Domestic', 'Non-Domestic'])
axes[0, 1].tick_params(axis='x', rotation=45)
axes[1, 0].axis('off')
axes[1, 1].axis('off')
plt.tight_layout()
plt.show()
```

From the First graph , we can observe that there were more arrests on the streets and more nonarrests in apartment areas. From the second graph we can observe that More number of domestic cases were in apartment areas and then residence areas.Most non domestic cases were found on the streets.

 Question 5 : What is the mean of the hourly total arrests count for the different months for three years combined.

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Hour'] = df['Date'].dt.hour
filtered_df = df[(df['Year'] >= 2021) & (df['Year'] <= 2023)]
monthly_hourly_arrests = filtered_df.groupby(['Month', 'Hour'])['Arrest'].mean().reset_index()

plt.figure(figsize=(15, 8))
sns.lineplot(data=monthly_hourly_arrests, x='Hour', y='Arrest', hue='Month', marker='o', palette='tab10')
plt.title('Mean Hourly Arrests Count for Different Months (2021-2023)')
plt.xlabel('Hour of the Day')
plt.ylabel('Mean Arrest Count')
plt.legend(title='Month', loc='upper right')
plt.grid(True)
plt.show()
```

We can observe that during the 3 rd. month there were high mean hourly arrests count at 18:00. We can also note that in the month of January there is a significant increase and decrease in the count.

Question 6 : What is the correlation between the different numerical features in the dataset?

```python
df['Arrest'] = df['Arrest'].astype(int)
df['Domestic'] = df['Domestic'].astype(int)
numerical_cols = df.select_dtypes(include=['float64', 'int64'])
corr_matrix = numerical_cols.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

We can observe that there is a negative correlation between arrest and domestic. Also there is a positive correlation between arrest and community area.

Question 7 : Is there a difference in crime rates during weekdays versus weekends and how it influences arrest rates?

```python
df['WeekdayOrWeekend'] = df['DayOfWeek'] >= 5
crime_counts = df['WeekdayOrWeekend'].value_counts()
arrest_rates = df.groupby('WeekdayOrWeekend')['Arrest'].mean()
fig, ax1 = plt.subplots(figsize=(10, 6))
ax1.bar(crime_counts.index, crime_counts.values, color='blue', alpha=0.5, label='Crime Count')
ax2 = ax1.twinx()
ax2.plot(arrest_rates.index, arrest_rates.values, color='red', marker='o', label='Arrest Rate')
ax1.set_xlabel('Day Type')
ax1.set_ylabel('Number of Crimes', color='blue')
ax2.set_ylabel('Arrest Rate', color='red')
ax1.set_xticks([0, 1])
ax1.set_xticklabels(['Weekday', 'Weekend'])
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')
plt.title('Crime Counts and Arrest Rates: Weekdays vs Weekends')
plt.show()
```

We can observe that the arrest rate is significantly higher in the weekends ranging to 0.1425 . The crime count is seen to be more in weekdays.

Question 8 : How does the frequency of different crime types vary across each community area?

```python
crime_counts = df.groupby(['Community Area', 'Primary Type']).size().reset_index(name='Count')
crime_counts_pivot = crime_counts.pivot(index='Community Area', columns='Primary Type', values='Count').fillna(0)
plt.figure(figsize=(24, 20))
crime_counts_pivot.plot(kind='bar', stacked=True, cmap='tab20', ax=plt.gca())
plt.title('Frequency of Crime Types Across Community Areas')
plt.xlabel('Community Area')
plt.ylabel('Frequency')
plt.legend(title='Crime Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The frequency of crime type is found to be more in community area 25.0.Frequency of Battery , theft and weapons violation seem to be more in most of the community areas.
